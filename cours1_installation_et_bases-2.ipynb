# Les variables

## int (nombres entiers)

### Présentation

Une variable est un nom auquel on associe une valeur, pour pouvoir la réutiliser plus tard sans la retaper. Le type `int` (integer) désigne les nombres entiers.

### Explication

Nous allons commencer par nommer une variable `a` et lui affecter la valeur 2.


```python
a = 2
```

Il s'est passé beaucoup de choses intéressantes en une seule ligne, de façon invisible à l'œil nu :
* Python a créé une variable qui s'appelle "a" ;
* il lui a réservé une place en mémoire RAM ;
* il l'a déclarée comme une variable numérique (int) ;
* il lui a affecté la valeur 2.

Python est beaucoup plus simple que d'autres langages pour faire la même chose. Voici, à titre de comparaison (pas besoin de comprendre), le code équivalent en C++ :


```python
int main()
{
    int a(2);
    return 0;
}
```

Ou bien en VBA :

```python
Sub main()

Dim a as Integer

a = 2

End Sub
```

### Exemple

Pour accéder au contenu de la variable `a`, on utilise la fonction `print` :


```python
print(a)
```

On peut faire des opérations mathématiques sur les variables numériques :

```python
a + 23
```

On peut aussi utiliser une variable pour en créer une autre. Le nom d'une variable ne doit pas commencer par un chiffre et ne doit pas contenir d'espace (on utilise des `_` à la place). Le nom doit être le plus explicite possible : un nom comme `a` est une mauvaise pratique en dehors d'un exemple pédagogique, car personne ne saura à quoi elle sert.

```python
autre_variable = a + 1
autre_variable
```

On peut aussi additionner deux variables entre elles :

```python
autre_variable + a
```

`a` et `autre_variable` stockent des `int` (nombres entiers). Pour vérifier le type d'une variable, on utilise la fonction `type` :

```python
type(a)
```

```python
type(autre_variable)
```

Si vous avez une erreur de type `NameError`, c'est que la variable que vous essayez d'utiliser n'a jamais été définie. Essayez, à titre d'exemple (l'erreur est normale ici) :

```python
variable_qui_n_existe_pas
```

Les floats sont également des variables numériques (on peut faire des calculs dessus), sauf qu'ils ont une partie décimale (avec un point `.` et non une virgule, contrairement au français).

Pour n'afficher que quelques chiffres après la virgule, on utilise la fonction `round(nombre, nombre_de_chiffres_apres_la_virgule)` :

```python
round(flottant, 2)
```

## str (chaînes de caractères / texte)

### Présentation

Le type `str` (string) désigne le texte. Pour déclarer du texte, on l'entoure de guillemets, simples `'` ou doubles `"` (peu importe lequel, du moment qu'on utilise le même des deux côtés).

### Explication et exemple


```python
'bonjour'
```

S'il n'y a pas de guillemets, Python va (à juste titre) comprendre que c'est le nom d'une variable, et va vouloir utiliser ce qu'elle contient. Essayez la cellule suivante (l'erreur `NameError` est normale) :

```python
bonjour
```

```python
c = 'bonjour'
type(c)
```

Python ne sait pas non plus additionner directement un `int` et un `str`. Essayez (l'erreur `TypeError` est normale) :

```python
f = "7"
e = a + f
```

En revanche, la fonction `print` sait afficher plusieurs éléments de types différents à la suite, séparés par des virgules :

```python
print('Le texte introduit le chiffre', f, a, 'bonjour')
```

Bon à savoir : l'opérateur `+` entre deux `str` va concaténer (coller) les deux chaînes de caractères :

```python
'text1' + 'text2'
```

## bool (booléens)

### Présentation

Les variables booléennes ne prennent que deux valeurs possibles : vrai (`True`) ou faux (`False`). Elles sont très utilisées pour les opérations logiques et les comparaisons.

### Explication et exemple


```python
bool_vrai = True
bool_faux = False
```

Les opérateurs logiques `or` (ou) et `and` (et) fonctionnent comme en logique classique :

```python
bool_vrai or bool_faux
```

```python
bool_vrai and bool_faux
```

Les booléens sont aussi le résultat de comparaisons, avec `==` (égal à) ou `!=` (différent de) :

```python
a == 10
```

```python
a == 3
```

```python
a != autre_variable
```

On peut aussi enchaîner des comparaisons :

```python
1 < 3 < 4
```

```python
3 < 2 < 5
```

## list (listes)

### Présentation

Une liste permet de regrouper plusieurs variables ensemble, comme un panier. Les éléments sont séparés par des virgules et regroupés entre crochets `[]`. Une liste peut contenir des éléments de types différents.

### Explication et exemple


```python
[1, 2, 'trois', a, autre_variable, c]
```

On peut stocker l'ensemble dans une variable pour la réutiliser :

```python
liste_de_variables = [1, 2, 'trois', a, autre_variable, c]
liste_de_variables
```

Notez que seul le **contenu** des variables (2, 3, 'bonjour'...) est conservé dans la liste, pas leur nom.

Le nombre d'éléments dans une liste s'obtient avec la fonction `len()` :

```python
len(liste_de_variables)
```

On accède à un élément d'une liste avec `liste[indice]`, l'indice étant la position de l'élément, **en commençant à compter à partir de 0** :

```python
liste_de_variables[0]
```

```python
liste_de_variables[3]
```

On peut extraire une sous-liste avec `liste[depart:arrivee]` :

```python
extract_de_liste = liste_de_variables[0:2]
print(extract_de_liste)
```

Pour ajouter un nouvel élément à une liste, on utilise la méthode `.append()` :

```python
extract_de_liste.append('nouvel élément')
extract_de_liste
```

Toutes les opérations possibles sur les listes sont référencées dans la documentation officielle : [https://docs.python.org/fr/3/tutorial/datastructures.html](https://docs.python.org/fr/3/tutorial/datastructures.html)
